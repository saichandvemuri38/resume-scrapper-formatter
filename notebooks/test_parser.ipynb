```python
import matplotlib.pyplot as plt
import numpy as np

# Example vectors
v1 = np.array([1, 2])  # "Career Highlights"
v2 = np.array([2, 3])  # "Work Experience"
v3 = np.array([2, -1]) # "Technical Skills"

# Cosine similarity formula
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("v1,v2 cosine:", cosine_similarity(v1, v2))
print("v1,v3 cosine:", cosine_similarity(v1, v3))

# Plot
plt.figure(figsize=(6,6))
origin = np.array([0,0])
plt.quiver(*origin, *v1, angles='xy', scale_units='xy', scale=1, color='r', label='Line')
plt.quiver(*origin, *v2, angles='xy', scale_units='xy', scale=1, color='g', label='Prototype Work Exp')
plt.quiver(*origin, *v3, angles='xy', scale_units='xy', scale=1, color='b', label='Prototype Skills')
plt.xlim(-1,3); plt.ylim(-2,4)
plt.grid(True)
plt.legend()
plt.title("Cosine Similarity Visual")
plt.show()
```
